# DataFrame of data from Reddit

```python
import glob
import pandas as pd

pd.set_option("display.max_colwidth", 60)   # keep long comments readable in the table

# all six subreddits, June 2018 comments
files = glob.glob(r"C:\reddit_parquet\comments\subreddit=*\year=2018\RC_2018-06.parquet")
df = pd.concat((pd.read_parquet(f) for f in files), ignore_index=True)

# readable date instead of a Unix timestamp
df["created"] = pd.to_datetime(df["created_utc"], unit="s", utc=True)

print(df.shape)          # (rows, columns)
df.info()                # column types, also nice for the report
df[["created", "subreddit", "author", "score", "body", "parent_id", "link_id"]].head(10)
```
