## DataFrame of datasets form Kaggle

```python
import pandas as pd
from pathlib import Path

DATA = Path("data")
dfs = {p.stem: pd.read_csv(p) for p in sorted(DATA.glob("*.csv"))}

for name, df in dfs.items():
    print(f"{name}: {df.shape}")
    display(df.head())
```

## Basic metrics

```python
summary = pd.DataFrame([
    {
        "dataset": name,
        "file_size_mb": (DATA / f"{name}.csv").stat().st_size / 1024**2,
        "memory_mb": df.memory_usage(deep=True).sum() / 1024**2,
        "rows": len(df),
        "columns": df.shape[1],
    }
    for name, df in dfs.items()
]).set_index("dataset")

summary.loc["TOTAL"] = summary.sum()
summary = summary.astype({"rows": "Int64", "columns": "Int64"}).round(2)
summary
```

## Frequency per category

Doomscrolling `Gender` is coded numerically (1 / 2 / 3); see the study codebook for labels.

```python
CATEGORY_COLS = {
    "Doomscrolling_Study2_Dataset": ["Gender"],
    "Doomscrolling_Study3_Dataset": ["Gender"],
    "Social_Media_Addiction": ["gender", "platform_usage", "social_interaction_level"],
    "reels_attention_span_dataset_12000": ["platform", "stress_level"],
    "sleep_doomscrolling_habits_clean": ["gender", "occupation_status", "sleep_quality_category", "doomscroller"],
    "smmh_clean": ["gender", "occupation_status", "avg_daily_time_social_media"],
    "student_sm_clean": ["gender", "academic_level", "purpose_of_use", "stress_level"],
}


def freq_table(df, cols):
    parts = []
    for col in cols:
        vc = df[col].value_counts(dropna=False)
        parts.append(
            pd.DataFrame({"count": vc, "%": (vc / len(df) * 100).round(1)})
            .rename_axis("value").reset_index().assign(column=col)
        )
    return pd.concat(parts).set_index(["column", "value"])


for name, cols in CATEGORY_COLS.items():
    print(f"{name} ({len(dfs[name])} rows)")
    display(freq_table(dfs[name], cols))
```
